# Apply Functions in Pandas

```python
def ab(x):
    if x< 36:
        return 'low'
    if x >=36 and x <=40:
        return 'moderate'
    else:
        return 'high'
```

```python
df['Temperature_Category'] = df['Temperature'].apply(ab)
```

```python
df
```

```python
df['Temperature_Category'].value_counts()
```
