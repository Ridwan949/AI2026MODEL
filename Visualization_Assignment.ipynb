## 1. Histogram
Shows the **distribution of one numerical variable** by splitting it into bins and counting how many values fall in each bin. Use it to see shape (skewed, symmetric), spread and outliers.

```python
plt.figure(figsize=(7,4))
sns.histplot(df["Age"], bins=20, color="steelblue")
plt.title("Distribution of Passenger Age")
plt.show()
```

## 2. Bar chart
Compares a **numerical value or count across categories**. Each bar's height is the count (or mean) of that category. Bars are separated because the categories are discrete.

```python
plt.figure(figsize=(7,4))
sns.countplot(x="Pclass", hue="Survived", data=df)
plt.title("Survival Count by Passenger Class")
plt.show()
```

## 3. Boxplot
Summarises a numerical variable with the **five-number summary**: minimum, Q1, median, Q3, maximum. Points beyond the whiskers are **outliers**. Great for comparing groups.

```python
plt.figure(figsize=(7,4))
sns.boxplot(x="Pclass", y="Fare", data=df)
plt.title("Fare by Passenger Class")
plt.show()
```

## 4. Scatterplot
Plots **two numerical variables** as points to reveal relationships, trends, clusters or outliers. Colour can add a third (categorical) variable.

```python
plt.figure(figsize=(7,5))
sns.scatterplot(x="Age", y="Fare", hue="Survived", data=df)
plt.title("Age vs Fare")
plt.show()
```

## 5. Pairwise plot (pair plot)
A grid of scatterplots for **every pair of numerical variables**, with histograms/densities on the diagonal. Quick overview of all relationships at once.

```python
sns.pairplot(df[["Survived", "Pclass", "Age", "Fare"]], hue="Survived", diag_kind="kde")
plt.show()
```

## 6. Pie chart
Shows the **proportion of each category** as slices of a circle. Best with few categories (≤5) that add up to a whole.

```python
counts = df["Embarked"].value_counts()
plt.figure(figsize=(5,5))
plt.pie(counts, labels=counts.index, autopct="%1.1f%%", startangle=90)
plt.title("Port of Embarkation (C = Cherbourg, Q = Queenstown, S = Southampton)")
plt.show()
```

## 7. Density chart (KDE)
A **smoothed version of the histogram**: a curve estimating the probability density of a numerical variable. Easier than histograms for comparing several groups on one plot.

```python
plt.figure(figsize=(7,4))
sns.kdeplot(data=df, x="Age", hue="Survived", fill=True, common_norm=False)
plt.title("Age Density by Survival")
plt.show()
```

## 8. Violin plot
Combines a **boxplot with a density curve**. The width at each value shows how many observations are there, so you see both summary stats and the full distribution shape.

```python
plt.figure(figsize=(7,5))
sns.violinplot(x="Pclass", y="Age", hue="Sex", split=True, data=df)
plt.title("Age by Class and Sex")
plt.show()
```
